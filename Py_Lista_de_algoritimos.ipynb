```python
lanche_str = input("Olá tudo bem quanto ta o lanche? ")
lanche = float(lanche_str.replace(',','.'))
bebida_str = input("E a bebida? ")
bebida = float(bebida_str.replace(',','.'))
resul_soma = lanche + bebida
resul_soma = str(resul_soma)
print("Tudo fica R$ " + resul_soma)
```

```python
idade1 = int(input("Eae Lucas qual sua idade: "))
idade2 = int(input("Olá Maria qual e sua idade: "))
idade3 = int(input("Eae Pedro qual sua idade: "))
soma_idade = idade1 + idade2 + idade3
media = soma_idade / 3
media = int(media)
print(f"A media de idade entre eles é de " + (str(media)) + " anos")
```

```python
altura1_str = str(input("Olá Lucas qual sua altura: "))
altura1 = float(altura1_str.replace(',','.'))
altura2_str = str(input("Olá Maria qual sua altura: "))
altura2 = float(altura2_str.replace(',','.'))
altura3_str = str(input("Eae Pedro qual sua altura: "))
altura3 = float(altura3_str.replace(',','.'))
soma_altura = altura1 + altura2 + altura3
media = soma_altura / 3
print(f"A media de altura entre eles é de {media:.2f} M")
```

```python
produto1_str = str(input("Olá Lucas qual o valor da coca-cola no mercado: "))
produto1 = float(produto1_str.replace(',','.'))
produto2_str = str(input("Olá Maria qual o valor da coca-cola no mercadinho da sua rua: "))
produto2 = float(produto2_str.replace(',','.'))
produto3_str = str(input("Eae Pedro qual o valor da coca-cola na adega: "))
produto3 = float(produto3_str.replace(',','.'))
soma_produto = produto1 + produto2 + produto3
media = soma_produto / 3
print(f"A media do preço da coca-cola entre os lugares de venda é de R$ {media:.2f}")
```

```python
peso1_str = str(input("Olá Lucas qual o seu peso: "))
peso1 = float(peso1_str.replace(',','.'))
peso2_str = str(input("Olá Maria qual o seu peso: "))
peso2 = float(peso2_str.replace(',','.'))
peso3_str = str(input("Eae Pedro qual o seu peso: "))
peso3 = float(peso3_str.replace(',','.'))
soma_peso = peso1 + peso2 + peso3
media = soma_peso / 3
print(f"A media do peso entre os lugares de venda é de {media:.2f} Kg")
```

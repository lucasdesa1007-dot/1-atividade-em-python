```python
jogos = ["Roblox", "Minecraft", "MR"]
jogos[2] = "COD" #Isso serve para que substitua um valor por outro.
jogos.append("MR") #O comando ".append server pra colocar algo la no final"
jogos.insert(0,"Overwatch") #O comando ".insert serve para colocar um item em uma determinda lugar."
jogos.remove("Minecraft") #O comando ".remove serve para remover um valor"
print(jogos)
if "Roblox" in jogos:
  print("Tem roblox nos jogos.")
```

```python
id = input("Informe qual seu ID de cadastro: ")
nome = input("Digite seu nome: ")
dados = {
    "id":id, "nome":nome
    }
print(dados)
#Agora serve para ver se o usuario quer trocar alguma informaçao
resposta_sim = "sim"
resposta = input("Deseja altera algo(sim/nao): ")
if resposta == resposta_sim:
 indice = input("Qual dado voce quer alterar? (id ou nome:)")
 novo_dado = input("Digite  novo dado: ")
 dados[indice] = novo_dado
 print(dados)
else:
  print("Cadastro mantido")

```
